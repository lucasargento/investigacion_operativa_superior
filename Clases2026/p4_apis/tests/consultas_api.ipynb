# ScheduleIt! Factory API — Consultas desde Python

Notebook para explorar los endpoints de la API.  
**Requisito:** tener el server corriendo en `localhost:8000`.

```bash
uvicorn server:app --reload --port 8000
```

````python
import requests
import pandas as pd

BASE_URL = "http://localhost:8000"
HEADERS = {"X-API-Key": "IO-TOKEN-GRP1-2025"}  # cambiar por tu token de grupo
````

## 1. Info general (sin auth)

````python
r = requests.get(f"{BASE_URL}/")
r.json()
````

## 2. Operarios

````python
# Traer todos los operarios (3 páginas de 10)
todos = []
for page in range(1, 4):
    r = requests.get(f"{BASE_URL}/operarios", headers=HEADERS, params={"page": page, "page_size": 10})
    data = r.json()
    todos.extend(data["data"])

df_operarios = pd.DataFrame(todos)
print(f"Total operarios: {len(df_operarios)}")
df_operarios
````

````python
# Filtrar operarios con habilidad 'A'
r = requests.get(f"{BASE_URL}/operarios", headers=HEADERS, params={"habilidad": "A", "page_size": 30})
pd.DataFrame(r.json()["data"])
````

````python
# Operario específico
r = requests.get(f"{BASE_URL}/operarios/5", headers=HEADERS)
r.json()
````

## 3. Lotes de Mezcla

````python
r = requests.get(f"{BASE_URL}/lotes/mezcla", headers=HEADERS, params={"page_size": 22})
data = r.json()
df_mezcla = pd.DataFrame(data["data"])
print(f"Total: {data['total_items']}")
df_mezcla
````

## 4. Lotes de Envasado

````python
r = requests.get(f"{BASE_URL}/lotes/envasado", headers=HEADERS, params={"page_size": 22})
data = r.json()
df_envasado = pd.DataFrame(data["data"])
print(f"Total: {data['total_items']}")
df_envasado
````

````python
# Solo urgentes
r = requests.get(f"{BASE_URL}/lotes/envasado", headers=HEADERS, params={"urgente": True, "page_size": 22})
pd.DataFrame(r.json()["data"])
````

## 5. Fechas de Entrega

````python
# Entrega de un lote específico
r = requests.get(f"{BASE_URL}/lotes/101/entrega", headers=HEADERS)
r.json()
````

````python
# Todas las entregas
r = requests.get(f"{BASE_URL}/entregas", headers=HEADERS)
data = r.json()
df_entregas = pd.DataFrame(data["entregas"])
print(f"Total: {data['total']}")
df_entregas
````

````python
# Entregas del lunes
r = requests.get(f"{BASE_URL}/entregas", headers=HEADERS, params={"dia": "lunes"})
pd.DataFrame(r.json()["entregas"])
````

## 6. Procesos Especiales

````python
r = requests.get(f"{BASE_URL}/procesos", headers=HEADERS)
pd.DataFrame(r.json()["procesos"])
````

````python
# Proceso específico
r = requests.get(f"{BASE_URL}/procesos/LIMP", headers=HEADERS)
r.json()
````

## 7. Manejo de errores

````python
# Token inválido → 401
r = requests.get(f"{BASE_URL}/operarios", headers={"X-API-Key": "TOKEN-MALO"})
print(f"Status: {r.status_code}")
r.json()
````

````python
# Operario inexistente → 404
r = requests.get(f"{BASE_URL}/operarios/999", headers=HEADERS)
print(f"Status: {r.status_code}")
r.json()
````
