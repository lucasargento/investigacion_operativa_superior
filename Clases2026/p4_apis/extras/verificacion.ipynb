# Verificación de conexión a la API

Seguí estos pasos para verificar que podés conectarte a la API de ScheduleIt!

## Paso 1: Configurar tu token y datos

```python
import requests

BASE_URL = "http://localhost:8000"  # Cambiar a la URL de Render cuando esté deployado

# --- COMPLETAR ---
TOKEN = "IO-2026-SCHEDULEIT"
NOMBRE = "Fernando"      # Tu nombre
APELLIDO = "Alonso"  # Tu apellido
# -----------------

HEADERS = {"X-API-Key": TOKEN}
```

## Paso 2: Obtener tu código secreto

Hacemos un `POST /verificar` enviando nombre y apellido. La API nos devuelve un código único.

```python
resp = requests.post(
    f"{BASE_URL}/verificar",
    headers=HEADERS,
    json={"nombre": NOMBRE, "apellido": APELLIDO},
)

print(f"Status: {resp.status_code}")
data = resp.json()
print(data)
if resp.status_code == 200:
    print("¡Verificación exitosa!")
    print(f"\nTu código: {data['codigo']}")
else:
    print("Error en la verificación.")
    print(f"Mensaje: {data['detail']}")
```

## (Bonus) Ver el ranking desde Python

```python
resp = requests.get(f"{BASE_URL}/verificaciones")
ranking = resp.json()

print(f"Total verificados: {ranking['total']}\n")
for r in ranking["ranking"]:
    print(f"  #{r['posicion']} - {r['nombre']} {r['apellido']} ({r['timestamp'][:19]})")
```
